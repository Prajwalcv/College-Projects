```python
import numpy as np
```

```python
#Holes Table Generation
def Holes_table_generator(holes_num):
  holes_table=[]
  current_address=0
  for i in range(holes_num):
    dic={}
    size=np.random.randint(1,5000)  # 5000 is the limit for hole size, you can change this limit if you want
    dic['start_address']=current_address
    dic["size(in mb)"]=size
    holes_table.append(dic)
    #No overlapping
    current_address+=size
  return holes_table
```

```python
#Memory Allocation
class Memory_Allocator:
    def __init__(self,holes_table):
        self.holes_table = holes_table
    def allocate_memory(self,mem_to_allocate):
      allocated=False
      for hole in self.holes_table:
        #If requested memory < available memory
        if hole["size(in mb)"]>mem_to_allocate:
          print("Allocated to memory with start address:",hole["start_address"])
          start_address=hole["start_address"]+mem_to_allocate
          hole["start_address"]=start_address
          remaining_mem=hole["size(in mb)"]-mem_to_allocate
          hole["size(in mb)"]=remaining_mem
          allocated=True
          break
        #If requested memory==available memory
        elif hole["size(in mb)"]==mem_to_allocate:
          print("Allocated to memory with start address:",hole["start_address"])
          self.holes_table.remove(hole)
          allocated=True
          break
      if allocated:
        print(f"\nAfter allocating memory of size {mem_to_allocate} mb:")
        self.show_table()
      else:
        print("Requested memory is greater than available memory")
        self.show_table()
    def show_table(self):
        if len(self.holes_table) == 0:
            print("No holes left")
            print('='*50)
            return
        print("="*50)
        print("Holes Table:")
        for i in self.holes_table:
          print(i,"\n")
        print('='*50)
```

```python
#Testing
tab_size=int(input("Enter number of holes:"))
hole_table=Holes_table_generator(tab_size)
ans=Memory_Allocator(hole_table)
ans.show_table()
while (True):
  n=int(input("\nEnter memory size to allocate:"))
  if n==0:
    break
  ans.allocate_memory(n)
  print("Enter number 0 to break")
```
